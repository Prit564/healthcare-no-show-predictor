```python
import pandas as pd
import numpy as np
df = pd.read_csv("cleaned_data.csv")
```

## We will simply analyze the range of each columns in this section

```python
df['no_show'].value_counts(normalize=True) * 100
```

From the data it is notable that only 8.26% of patients did not show up in the following date whereas 91.74% patients did show up. This implies that most of the patients tend to follow the doctor's advice (almost 9 out of 10). We will analyze why the 10% of patients do otherwise in the upcoming files.

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(10, 6))
sns.histplot(data=df, x='age', bins=30, kde=True, color='skyblue')
plt.title('Age Distribution of Patients')
plt.xlabel('Age')
plt.ylabel('Number of Patients')
plt.show()
```

The histogram indicates that the majority of patients fall between the ages of 30 and 60, with the distribution peaking roughly around age 40. There is also a significant, distinct spike at age 0, representing a high volume of pediatric appointments for infants. After age 60, the number of appointments gradually declines.

```python
plt.figure(figsize=(10, 6))
sns.histplot(data=df, x='distance_km', bins=30, kde=True, color='skyblue')
plt.title('Distance of Hospital from Patients Home')
plt.xlabel('Distance(km)')
plt.ylabel('Number of Patients')
plt.show()
```

The histogram for travel distance is heavily right-skewed. The vast majority of patients live within a close radius of the hospital (between 0 and 10 km). As the distance increases beyond 15 km, the number of patients drops off significantly, though there are a few outliers who travel up to 50 km or more.

```python
plt.figure(figsize=(10, 6))
sns.histplot(data=df, x='waiting_days', bins=30, kde=True, color='skyblue')
plt.title('Waiting Days Before Next Date')
plt.xlabel('Days')
plt.ylabel('Counts')
plt.show()
```

The distribution of waiting days is heavily right-skewed. This indicates that the majority of patients have a relatively short wait time between scheduling their visit and the actual appointment date. The volume of appointments peaks sharply between 5 and 15 days, and then declines steadily. While some patients wait 40 days or more, these represent a very small fraction of the total dataset.

```python
plt.figure(figsize=(10, 6))
sns.countplot(data=df, x='income_level', order=df['income_level'].value_counts().index)
plt.title('Patient Distribution by Income Level')
plt.xlabel('Income Level')
plt.ylabel('Number of Patients')
plt.xticks(rotation=45)
plt.show()
```

The patient base is predominantly comprised of middle and lower-middle-income individuals, with both groups noticeably exceeding 10,000 patients. The low and upper-middle-income groups represent a moderate portion of the data (nearly equal in size), while high-income patients make up the smallest segment of the clinic's population.

```python
plt.figure(figsize=(10, 6))
sns.countplot(data=df, x='marital_status', order=df['marital_status'].value_counts().index)
plt.title('Patient Distribution by Marital Status')
plt.xlabel('Marital Status')
plt.ylabel('Number of Patients')
plt.xticks(rotation=45)
plt.show()
```

The marital status distribution shows that married individuals represent the largest segment of the clinic's patient population by a wide margin, exceeding 25,000 appointments. Single patients make up the second-largest group, while divorced and widowed individuals constitute relatively small fractions of the total dataset.
