### All the columns will not be necessary. We are removing too weak columns for our analysis.

```python
df = df.drop(columns = ["appointment_id","city","day_of_week","insurance_type","education_level"])
```

```python
df
```

```python
# 1. Converting categorical text to Category type
cat_cols = [
    'patient_id', 'gender', 'neighborhood', 'transportation_type', 
    'appointment_type', 'doctor_specialty', 'clinic_type', 
    'weather_condition', 'employment_status', 'income_level', 'marital_status'
]
df[cat_cols] = df[cat_cols].astype('category')

# 2. Converting date strings to Datetime type
df['scheduled_day'] = pd.to_datetime(df['scheduled_day'])
df['appointment_day'] = pd.to_datetime(df['appointment_day'])

# 3. Optimizing integers (downcast sizes to save memory)
int_cols = [
    'sms_received', 'has_chronic_disease', 'hypertension', 
    'diabetes', 'alcoholism', 'disability', 'rainy_day', 'no_show', 'age', 'waiting_days', 'appointment_hour', 'month', 
    'previous_appointments', 'previous_no_shows', 'children_count', 'reminder_calls'
]
for col in int_cols:
    df[col] = pd.to_numeric(df[col], downcast='integer')

# 4. Optimizing floats (downcast sizes to save memory)
float_cols = ['no_show_rate', 'distance_km', 'temperature_celsius', 'no_show_probability']
for col in float_cols:
    df[col] = pd.to_numeric(df[col], downcast='float')

# Using nullable Float64 for satisfaction score due to missing data (NaN)
df['patient_satisfaction_score'] = df['patient_satisfaction_score'].astype('Float64')
```

```python
df.isnull().sum()
```

Look, 2000 data missing in the column "income_level". we will drop the rows with these missing data.

```python
df = df.dropna(subset = "income_level")
```
